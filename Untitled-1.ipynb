```python
df['Year'] = df['Order Date'].dt.year
df['Month_Name'] = df['Order Date'].dt.month_name()
df['Day'] = df['Order Date'].dt.day_name()
```

```python
df
```

```python
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100
df[['Sales', 'Profit', 'Profit Margin']].head()
```

```python
df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days
df[['Order Date', 'Ship Date', 'Shipping Duration']].head()
```

```python
print("Average Shipping Duration:", df['Shipping Duration'].mean())
```

```python
def sales_category(sales):
    if sales < 100:
        return 'Low'
    elif sales < 500:
        return 'Medium'
    else:
        return 'High'

df['Sales Performance Category'] = df['Sales'].apply(sales_category)

df[['Sales', 'Sales Performance Category']].head()
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

sales_outliers = df[
    (df['Sales'] < lower) |
    (df['Sales'] > upper)
]

print("Number of Sales Outliers:", len(sales_outliers))
```

```python
Q1 = df['Profit'].quantile(0.25)
Q3 = df['Profit'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

profit_outliers = df[
    (df['Profit'] < lower) |
    (df['Profit'] > upper)
]

print("Number of Profit Outliers:", len(profit_outliers))
```

```python
def generate_summary(df):

    summary = {
        "Total Sales": df["Sales"].sum(),
        "Total Profit": df["Profit"].sum(),
        "Total Orders": df["Order ID"].nunique(),
        "Total Customers": df["Customer ID"].nunique(),
        "Average Sales": df["Sales"].mean(),
        "Average Profit": df["Profit"].mean()
    }

    return summary
```

```python
summary = generate_summary(df)

for key, value in summary.items():
    print(key, ":", round(value, 2))
```

```python
def generate_kpi_summary(df):

    kpi = {
        "Total Sales": df['Sales'].sum(),
        "Total Profit": df['Profit'].sum(),
        "Total Orders": df['Order ID'].nunique(),
        "Total Customers": df['Customer ID'].nunique(),
        "Average Order Sales": df['Sales'].mean(),
        "Average Order Profit": df['Profit'].mean(),
        "Average Shipping Days": df['Shipping Duration'].mean(),
        "Average Profit Margin": df['Profit Margin'].mean()
    }

    return kpi
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=profit_by_region.index,
    y=profit_by_region.values
)

plt.title('Total Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit')

plt.show()
```

```python
kpi = generate_kpi_summary(df)

for key, value in kpi.items():
    print(key, ":", round(value, 2))
```
